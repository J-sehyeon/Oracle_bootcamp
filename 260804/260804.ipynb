```python
from typing import Annotated
from pydantic import BaseModel, Field

class NewUser(BaseModel):
    # Annotated를 사용하여 자료형(str)과 조건(min_length=3)을 명확하게 분리
    # 가방(str)에 '최소 3글자'라는 이름표(Field)를 붙였다고 이해하시면 됩니다.
    사는곳: Annotated[str, Field(min_length=10)]
    
    # 나이는 정수(int)이며, 0 이상 120 이하라는 조건을 추가
    age: Annotated[int, Field(ge=0, le=120)]


NewUser(사는곳="서울시 동작구 대방동 여성플라자", age=25)
```
