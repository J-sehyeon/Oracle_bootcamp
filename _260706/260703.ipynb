```python
def count_down(n):
    while n > 0:
        yield n  # 값을 하나씩 반환, 함수 상태 유지
        n -= 1


for num in count_down(5):
    print(num)  # 5, 4, 3, 2, 1
```

```python
def fibo_generator():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b
```

```python
fibo = fibo_generator()
```

```python
import io
import requests
import pandas as pd
```

```python
url = 'https://finance.naver.com/marketindex/exchangeDailyQuote.naver?marketindexCd=FX_USDKRW&page=1'
```

```python
df = pd.concat([pd.read_html(io.StringIO(requests.get(url.format(x)).text))[0] for x in range(1, 4)])
df.head()
```

```python
df.reset_index(drop=True, inplace=True)
df.to_json()

```

```python
prompt_template = PromptTemplate(
    template="""당신은 경제 전문가입니다. 지금 제공받은 환율 정보를 바탕으로 분석을 하세요.
        {usd} \n\n\n
        {subject}
                    """,
    input_variables=['subject'],
    partial_variables={'usd': df.to_json()}
)

chain = prompt_template | llm | StrOutputParser()
```

```python
chain.invoke({"subject" : "최근 환율이 어떻게 되어가는지 추이를 말해봐"})
```
