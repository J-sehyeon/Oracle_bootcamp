```python
df1_grp = df1.groupby(['날짜', '호선', '역명', '구 분'], as_index=False).sum().drop('할인', axis=1)
```

```python
df1_grp
```

```python
df1_grp_drop = df1_grp.drop(['04 ~ 05', '01 ~ 02', '02 ~ 03', '03 ~ 04'], axis=1)
```

```python
df1_grp_drop
```

```python
import re

p = re.compile('([가-힣0-9]+)\(')
```

```python
df1_grp_drop['지하철역명'] = df1_grp_drop.역명.apply(lambda x : p.findall(x)[0])
```

```python
df1_grp_drop = df1_grp_drop[['날짜', '호선', '구 분', '지하철역명', '05 ~ 06', '06 ~ 07', '07 ~ 08', '08 ~ 09',
       '09 ~ 10', '10 ~ 11', '11 ~ 12', '12 ~ 13', '13 ~ 14', '14 ~ 15',
       '15 ~ 16', '16 ~ 17', '17 ~ 18', '18 ~ 19', '19 ~ 20', '20 ~ 21',
       '21 ~ 22', '22 ~ 23', '23 ~ 24', '00 ~ 01']].copy()
```

```python
df1_grp_final = df1_grp_drop.groupby(['날짜', '호선', '구 분', '지하철역명'], as_index=False).sum()
```

```python
df1_grp_final
```

```python
df1_grp_final.columns
```

```python
df1_grp_final.columns = ["".join(x.split()) for x in df1_grp_final.columns]
```

```python
df1_grp_final.drop('호선', axis=1, inplace=True)
```

```python
df2 = df2[['날짜', '구분', '역명', '05~06', '06~07', '07~08', '08~09', '09~10', '10~11',
       '11~12', '12~13', '13~14', '14~15', '15~16', '16~17', '17~18', '18~19',
       '19~20', '20~21', '21~22', '22~23', '23~24', '00~01', '01~02']].copy()
```

```python
df2.rename(columns={"역명" : "지하철역명"}, inplace=True)
```

```python
df2['지하철역명'] = df2.지하철역명.apply(lambda x : p.findall(x)[0])
```

```python
df2_final = df2.groupby(['날짜', '구분', '지하철역명'], as_index=False).sum()
```

```python
df2_final.drop("01~02", axis=1, inplace=True)
```

```python
df3.columns = [x.replace('시', '') for x in df3.columns]
```

```python
df3.columns
```

```python
df3 = df3[['날짜', '구분', '역명', '05~06', '06~07', '07~08', '08~09', '09~10', '10~11',
       '11~12', '12~13', '13~14', '14~15', '15~16', '16~17', '17~18', '18~19',
       '19~20', '20~21', '21~22', '22~23', '23~24', '00~01']].copy()
```

```python
df3.rename(columns={'역명' : '지하철역명'}, inplace=True)
```

```python
df3
```

```python
df3_final = df3.groupby(['날짜', '구분', '지하철역명'], as_index=False).sum()
```

```python
print(df3_final.shape)
print(df2_final.shape)
print(df1_grp_final.shape)
```

```python
subway_2016 = pd.concat([df1_grp_final, df2_final, df3_final], ignore_index=True)
```

```python
subway_2016
```

```python
subway_2016['총합'] = subway_2016.iloc[:, 3:].sum(axis=1)
```

```python
subway_2016.to_csv('../data/subway.csv')
```
