```python
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
from langchain.tools import tool
import os
import pandas as pd 
import io 
import requests 
from langgraph.prebuilt import create_react_agent



load_dotenv()

api_key = os.getenv("OLLAMA_API_KEY")
openai_api_key = os.getenv('OPENAI_API_KEY')
headers = {
   "Authorization": f"Bearer {api_key}"
}
llm = ChatOllama(
   base_url="https://ollama.com", # 원격 서버 주소
   model="gemma4:31b-cloud",
   client_kwargs={"headers": headers},
   temperature=0.2,
   reasoning=True
)
```

```python
openai_llm = ChatOpenAI(model='gpt-5.4-mini-2026-03-17', api_key=openai_api_key)
```

```python
prompt = PromptTemplate(
    template="""30년차 전문 개발자입니다. 아래 제시된 코딩 테스트 문제에 대해서 
     코딩이 처음인 친구를 위해서 필요한 지식을 차례 차례 설명하세요 \n\n {problem}""",
    input_variables=['problem']
)
chain = prompt | llm | StrOutputParser()
rt = chain.invoke({'problem' : text})
```

```python
from langchain_core.prompts import ChatPromptTemplate
from pydantic import BaseModel, Field
from langchain_core.output_parsers import PydanticOutputParser
```

```python
CoteMaster = """ 
아래 제시된 문제를 Python 코드로 풀고, 왜 이렇게 작성했는지 설명 할 것
{format_instruction}
""" 
```

```python
class Answer(BaseModel):
    python_code : str = Field(description="제시된 문제에서 파이썬 코드를 작성하여 여기에 저장")
    answer : str = Field(description="제시된 파이썬 코드에 대한 자세한 설명을 한다.")
    
parser = PydanticOutputParser(pydantic_object=Answer)
# 1. 프롬프트 템플릿 정의
prompt = ChatPromptTemplate.from_messages([
    ("system", CoteMaster),
    ("human", '{problem}')
]).partial(
    format_instruction = parser.get_format_instructions()
)

cote_chain = prompt | openai_llm | parser
```

```python
cote_result = cote_chain.invoke({'problem' : text})
```

```python
print(cote_result.answer)
```

```python
with open("./tmp_code.py", "w") as f:
    f.write(cote_result.python_code)
rt = subprocess.run('python tmp_code.py', shell=True, text=True, capture_output=True)


print(rt.stdout)
```
