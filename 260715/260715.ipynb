```python
import re
def has_apologized(text: str) -> bool:
    """
    모델의 발언에 명시적인 사과 표현이 포함되었는지 확인한다.
    """
    apology_patterns = [
        r"미안해",
        r"미안하다",
        r"미안했어",
        r"미안합니다",
        r"죄송해",
        r"죄송하다",
        r"죄송합니다",
        r"사과할게",
        r"사과한다",
        r"내가\s*잘못했어",
        r"제\s*잘못",
    ]


    return any(
        re.search(pattern, text, flags=re.IGNORECASE)
        for pattern in apology_patterns
    )
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage
```

```python
def build_message_history(
    persona: str,
    current_role: str,
    conversation_history: list[dict[str, str]],
) -> list:
    """
    현재 발언자의 기존 발언은 AIMessage로 변환하고,
    상대방의 발언은 HumanMessage로 변환한다.


    에이전트마다 자신의 발언과 상대방의 발언을
    올바르게 구분하도록 대화 기록을 구성한다.
    """
    messages = [
        SystemMessage(content=persona),
        HumanMessage(content=상황_설명),
    ]


    for utterance in conversation_history:
        speaker = utterance["speaker"]
        content = utterance["content"]


        if speaker == current_role:
            messages.append(
                AIMessage(content=content)
            )
        else:
            messages.append(
                HumanMessage(
                    content=f"{speaker}의 발언:\n{content}"
                )
            )


    # 대화 기록이 없으면 첫 발언을 요청한다.
    if not conversation_history:
        messages.append(
            HumanMessage(
                content=(
                    "대화를 시작해 주세요. "
                    "현재 느끼는 감정과 서운한 이유를 이야기하세요."
                )
            )
        )


    return messages
```

```python
MAX_TURNS = 10

대화_기록: list[dict[str, str]] = []
대화_종료 = False
```

```python
여자_문장 = ""
for turn in range(1, MAX_TURNS + 1):
    print(f"\n===== {turn}번 턴 =====")


    # -----------------------------------------------------
    # 남자친구 차례
    # -----------------------------------------------------
    남자_messages = build_message_history(
        persona=남자_페르소나,
        current_role="남자",
        conversation_history=대화_기록,
    )


    남자반응 = male_agent.invoke(남자_messages)
    남자_문장 = 남자반응.text.strip()


    대화_기록.append(
        {
            "speaker": "남자",
            "content": 남자_문장,
        }
    )


    print(f"[남자] {남자_문장}")


    # 남자가 먼저 사과한 경우
    if has_apologized(여자_문장):
        대화_종료 = True
        break


    # -----------------------------------------------------
    # 여자친구 차례
    # -----------------------------------------------------
    여자_messages = build_message_history(
        persona=여자_페르소나,
        current_role="여자",
        conversation_history=대화_기록,
    )


    여자반응 = female_agent.invoke(여자_messages)
    여자_문장 = 여자반응.text.strip()


    대화_기록.append(
        {
            "speaker": "여자",
            "content": 여자_문장,
        }
    )


    print(f"[여자] {여자_문장}")


    # 여자가 먼저 사과한 경우
    if has_apologized(남자_문장):
        대화_종료 = True
        break
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableBranch
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field
```

```python
from langchain_ollama import ChatOllama

llm_local = ChatOllama(
    model="gemma4:31b-mlx",
    temperature=0.2,
    reasoning=True,
)
```

```python
class RouteClassification(BaseModel):
    category: str = Field(description="사용자가 입력한 질문의 내용을 파악해서 '연애', '진로' 둘 중 하나의 값을 리턴해야 한다.")
parser = PydanticOutputParser(pydantic_object=RouteClassification)
```

```python
classifier_prompt = PromptTemplate(
    template="다음 텍스트의 주제를 분석하여 카테고리를 분류하십시오.\n\n{format_instructions}\n\n어떠한 부가 설명도 하지 마십시오.\n\n텍스트: {text}",
    input_variables=["text"],
    partial_variables={"format_instructions": parser.get_format_instructions()},
)
classifier_chain = classifier_prompt | llm_local | parser
```

```python
연애 = PromptTemplate.from_template("""
    당신은 연애심리학, 애착이론, 커뮤니케이션 이론에 기반하여 상담하며, 항상 공감 40%, 분석 40%, 실천 조언 20% 비율로 답한다.
    {question}""") | llm_local

진로 = PromptTemplate.from_template("""
    당신은 진로심리학, 노동시장 동향, 역량 기반 커리어 설계, 강점 코칭 이론을 활용하여 상담하며,
    사용자의 적성(40%), 시장성(30%), 성장 가능성(20%), 삶의 만족도(10%)를 종합적으로 고려하여 조언한다.
    {question}""") | llm_local


일반 = PromptTemplate.from_template("""
    일반 상담사입니다.
    {question}""") | llm_local
```

```python
branch = RunnableBranch(
    (lambda x : x['classification'].category == '연애', 연애),
    (lambda x : x['classification'].category == '진로', 연애),
    일반
)
```

```python
from langchain_core.runnables import RunnablePassthrough

workflow = (
    {
        'question': RunnablePassthrough(),
        'classification': classifier_chain
    }
    | branch
)
```
