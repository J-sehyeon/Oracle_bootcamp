```python
import os
from openai import OpenAI
import base64
def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")
base64_image = encode_image("image.png")


client = OpenAI()
response = client.chat.completions.create(
    model="gpt-5.4-2026-03-05",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "text",
                    "text": "이 이미지에 포함된 모든 텍스트를 있는 그대로 정확하게 추출해 주세요.markdown를 사용해서 사용자가 보기 편하게 표로 작성할 것"
                },
                {
                    "type": "image_url",
                    "image_url": {
                        "url": f"data:image/png;base64,{base64_image}"
                    }
                }
            ]
        }
    ])


res = response.choices[0].message.content
```

```python
import markdown
print(markdown.markdown(res, extensions=['tables']))

lunch = markdown.markdown(res, extensions=['tables'])
```
