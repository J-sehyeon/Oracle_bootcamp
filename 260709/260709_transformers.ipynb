```python
import torch

device = "mps" if torch.mps.is_available() else "cpu"
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
```

```python
model_name = "WhitePeak/bert-base-cased-korean-sentiment"
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2)
```

```python
model
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = model.to('mps')
```

```python
def predict_sentiment(text):
    model.eval()
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=128,
        padding=True
    ).to(device)
    with torch.no_grad():
        outputs = model(**inputs)


    logits = outputs.logits
    probs = torch.softmax(logits, dim=-1)
    prediction = torch.argmax(probs, dim=-1).item()
   
    return prediction, probs[0]
```

```python
text = "오늘 비가 와서 정말로 좋을까안좋을까좋을까안좋을까좋을까안좋을까좋을까안좋을까좋을까안좋을까좋을까안좋을까."
label, score = predict_sentiment(text)

print(f'text: {text} \nlabel: {label} | score: {score}')
```
